### Display number of images and unique number of patients for the Raw Dataset

## Keep only metadata for working dataset

```python
# Verify the existence of image files
image_dir = "Data/working dataset/images"

existing_images = [f for f in os.listdir(image_dir) if os.path.isfile(os.path.join(image_dir, f))]
print("Number of existing image files:", len(existing_images))
```

```python
df_subset = df[df['Image Index'].isin(existing_images)]
```

### Display number of images and unique number of patients for the Raw Dataset

```python
print("Number of records in the subset:", len(df_subset))
print("Number of unique patients:", df_subset['Patient ID'].nunique())
df_subset.head()
```

### Data Quality Check

```python
#data overview
df_subset.info()
```

```python
#Missing values
missing_values = df_subset.isnull().sum()
print("Missing values per column:")
print(missing_values)
```

```python
#Duplicate Records
duplicate_records = df_subset[df_subset.duplicated()]
print("Number of duplicate records:", len(duplicate_records))
```

```python
#count unique labels
all_labels= set()
for labels in df_subset['Finding Labels']:
    all_labels.update(labels.split('|'))

print("Number of unique labels:", len(all_labels))
print(sorted(all_labels))
```

```python
#transform multi-labels into binary columns
all_labels = set()

for labels in df_subset["Finding Labels"]:
    all_labels.update(labels.split("|"))

diseases = sorted(all_labels)

for disease in diseases:
    df_subset[disease] = df_subset["Finding Labels"].apply(
        lambda x: 1 if disease in x.split("|") else 0
    )
```

```python
#count the number of diseases per image
df_subset["Disease_Count"] = (
    df_subset["Finding Labels"]
    .str.split("|")
    .apply(len)
)
```

```python
print("Rows:", df_subset.shape[0])
print("Columns:", df_subset.shape[1])

print("\nGender Distribution:")
print(df_subset["Patient Sex"].value_counts())

print("\nView Positions:")
print(df_subset["View Position"].value_counts())
```

```python
df_subset.head(5)
```

### Patient Age Distribution

```python
plt.figure(figsize=(10,6))
df_subset["Patient Age"].hist(bins=30)

plt.title("Patient Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

### Disease Frequency

```python
disease_counts = {}

for disease in diseases:
    disease_counts[disease] = df_subset[disease].sum()

disease_counts = pd.Series(disease_counts)

disease_counts.sort_values(
    ascending=False
).plot(
    kind='bar',
    figsize=(12,6)
)

plt.title("Disease Frequency")
plt.ylabel("Number of Images")
plt.show()
```

### Most common diseases

```python
common_diseases = disease_counts.sort_values(ascending=False).head(10).index.tolist()
print("Top 10 most common diseases:", common_diseases)
```

### Co-occurant diseases

```python
#disease combinations
combos= Counter(df_subset["Finding Labels"].apply(lambda x: tuple(sorted(x.split("|")))))
print("Top 10 most common disease combinations:", combos.most_common(10))
```

### Occurence by Gender

```python
#disease occurenance by gender

g_disease_counts = df_subset.groupby("Patient Sex")[diseases].sum()
g_disease_counts.T.plot(
    kind='bar', 
    figsize=(12,6)
)
plt.title("Disease Occurrence by Gender")
plt.ylabel("Number of Images")
plt.xlabel("Diseases")
plt.legend(title="Gender", loc='upper right')
plt.show()
```

### Visualisation

### Preview images with diagnoses

```python
sample_df = df_subset[
    df_subset["Finding Labels"] != "No Finding"
].sample(9, random_state=42)

fig, axes = plt.subplots(3, 3, figsize=(15, 15))

for ax, (_, row) in zip(axes.flatten(), sample_df.iterrows()):
    
    img_path = os.path.join(image_dir, row["Image Index"])
    
    img = Image.open(img_path)

    ax.imshow(img, cmap="gray")

    # Show diagnosis as title
    ax.set_title(
        row["Finding Labels"],
        fontsize=8
    )

    ax.axis("off")

plt.tight_layout()
plt.show()
```

## Healthy vs Diseased Images

```python
healthy_img=df_subset[df_subset["Finding Labels"] == "No Finding"].iloc[0]
diseased_img=df_subset[df_subset["Finding Labels"] != "No Finding"].iloc[0]
```

```python
##display both images side-by-side

fig, axes = plt.subplots(1, 2, figsize=(18, 9))
for i,row in enumerate([healthy_img, diseased_img]):
    img_path = os.path.join(image_dir, row["Image Index"])
    img = Image.open(img_path)
    axes[i].imshow(img, cmap="gray")
    axes[i].set_title(
        row["Finding Labels"],
        fontsize=12
    )
    axes[i].axis("off")
plt.show()
```
